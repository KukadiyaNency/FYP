***EDA: Exploratory Data Analysis***


> ***Exploratory Data Analysis (EDA) is a crucial step in understanding the underlying patterns, trends, and distributions within the dataset before applying formal statistical methods.***

> ***In this notebook, I conduct sector-wise EDA on data collected from the Office for National Statistics (ONS), focusing on five key sectors: Hospitality, Retail, Manufacturing, Transport and Storage, and Information and Communication.***

> ***The goal is to explore variables related to financial performance, workforce dynamics, trade variations, and strategic adaptations during the COVID-19 period. This includes analyzing trends over time, visualising key metrics, and preparing the data for hypothesis testing.***


> ***This notebook contains codes and results of 'Data preprocessing' and 'Exploratory data analysis (EDA)'***

> ***Each block of the code represents the names with initials of particular sector.***

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Hospitality Financial (HF)

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
# Loading the CSV file.

HF = pd.read_csv('/content/drive/MyDrive/Raw_fyp/Hospitality_Financial_Perfomance.csv')
```

```python
HF
```

```python
# 1. Standardising Date Format

HF.iloc[:, 0] = pd.to_datetime(HF.iloc[:, 0], dayfirst=True, errors='coerce').dt.strftime('%d/%m/%Y')
```

```python
HF
```

```python
# 2. Replacing '*' with NaN

HF.replace('*', np.nan, inplace=True)
```

```python
HF
```

```python
# 3. Replacing spaces with underscores in column names

HF.columns = HF.columns.str.replace(" ", "_")
```

```python
HF.dtypes
```

```python
# 4. Converting percentage columns to float

cols_to_convert = HF.columns[1:]  # Have to exclude 'Date' column
HF[cols_to_convert] = HF[cols_to_convert].replace('%', '', regex=True).astype(float)
```

```python
HF.dtypes
```

```python
# 5. getting the percentage of missing values in each column.

missing_pct = round(HF.isnull().sum()/len(HF)*100,1)
print(missing_pct)
```

```python
HF
```

```python
HF.ffill(inplace=True)  # First, try forward fill
HF.bfill(inplace=True)  # Then, backward fill as a backup
HF.interpolate(method='linear', inplace=True)  # Finally, interpolate remaining gaps
```

```python
HF
```

```python
HF.isnull().sum()
```

```python
# 6. Save new cleaned file

cleaned_file_path = "/content/HF_cleaned.csv"
HF.to_csv(cleaned_file_path, index=False)
```

```python
HF = pd.read_csv("/content/drive/MyDrive/Cleaned_FYP/HF_cleaned.csv")
```

```python
HF.describe()
```

```python
# Set 'Date' as index...
HF.set_index("Date", inplace=True)

# Plotting the trends...
plt.figure(figsize=(12, 6))
for col in HF.columns:
    plt.plot(HF.index, HF[col], label=col)

plt.xlabel("Date")
plt.ylabel("Turnover Percentage")
plt.title("Turnover Trends Over Time")
plt.xticks(rotation=45)
plt.grid(True)
plt.legend()
plt.show()
```

```python
HF.head(5)
```

```python
HF["Turnover_Increased"] = HF.iloc[:, 0:3].sum(axis=1)
HF["Turnover_Decreased"] = HF.iloc[:, 4:7].sum(axis=1)

plt.figure(figsize=(12, 6))
plt.plot(HF.index, HF["Turnover_Increased"], label="Increased Turnover", color="green")
plt.plot(HF.index, HF["Turnover_Unaffected"], label="Unaffected Turnover", color="blue")
plt.plot(HF.index, HF["Turnover_Decreased"], label="Decreased Turnover", color="red")

plt.xlabel("Date")
plt.ylabel("Percentage")
plt.title("Grouped Turnover Trends")
plt.xticks(rotation=45)
plt.legend()
plt.grid()
plt.show()
```

```python
# Setting Date as index
TS.set_index("Date", inplace=True)

# Plotting trends
plt.figure(figsize=(12, 6))
for col in TS.columns:
    plt.plot(TS.index, TS[col], label=col)

plt.xlabel("Date")
plt.ylabel("Percentage of Businesses")
plt.title("Transportation Strategic Trends Over Time")
plt.legend(loc="upper left", bbox_to_anchor=(1,1))
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
# Summing up relevant columns for grouped insights
TS["Using_Schemes"] = TS["Job_Retention_Bonus"] + TS["Kickstart_Job_Scheme_for_young_people"]
TS["Not_Sure"] = TS["Not_sure_"]
TS["Not_Using_Schemes"] = TS["Not_using,_or_intending_to_use,_either_of_these_schemes"]

# Plotting grouped import trends
plt.figure(figsize=(12,6))
plt.plot(TS.index, TS["Using_Schemes"], label="Schemes Using", color="red")
plt.plot(TS.index, TS["Not_Sure"], label="Not Sure", color="blue")
plt.plot(TS.index, TS["Not_Using_Schemes"], label="Not Using Any Schemes", color="green")

plt.xlabel("Date")
plt.ylabel("Strategic use Percentage")
plt.title("Grouped Transportation Strategic Trends")
plt.xticks(rotation=45)
plt.legend()
plt.grid()
plt.show()
```

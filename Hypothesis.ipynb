```python
# Split the dataset into two periods
early_financial_period = df_financial[df_financial['Date'] < '2021-01-01']
late_financial_period = df_financial[df_financial['Date'] >= '2021-01-01']

early_period_workforce = df_workforce[df_workforce['Date'] < '2021-01-01']
late_period_workforce = df_workforce[df_workforce['Date'] >= '2021-01-01']

early_period_import = df_trade_import[df_trade_import['Date'] < '2021-01-01']
late_period_import = df_trade_import[df_trade_import['Date'] >= '2021-01-01']

early_period_export = df_trade_export[df_trade_export['Date'] < '2021-01-01']
late_period_export = df_trade_export[df_trade_export['Date'] >= '2021-01-01']

early_period_strategic = df_strategic[df_strategic['Date'] < '2021-01-01']
late_period_strategic = df_strategic[df_strategic['Date'] >= '2021-01-01']
```

```python
# Function to perform T-Test on two selected columns from early vs late period
def perform_ttest(early_df, late_df, columns):
    results = []
    for col in columns:
        t_stat, p_val = ttest_ind(early_df[col].dropna(), late_df[col].dropna(), equal_var=False)
        results.append({
            'Column': col,
            'T-Statistic': t_stat,
            'P-Value': p_val
        })
    return pd.DataFrame(results)
```

```python
# Step 2: Reusable T-Test Function
def perform_ttest(early_df, late_df, columns):
    results = []
    for col in columns:
        t_stat, p_val = ttest_ind(early_df[col].dropna(), late_df[col].dropna(), equal_var=False)
        results.append({'Column': col, 'T-Statistic': t_stat, 'P-Value': p_val})
    return pd.DataFrame(results)

# Step 3: Define 2 columns to test per sector
hospitality_columns_v1 = [
    'Hospitality_Turnover_Has_Decreased_By_Up_To_20%',
    'Hospitality_Turnover_Unaffected'
]

retail_columns_v1 = [
    'Retail_Exporting_has_not_been_affected',
    'Retail_The_business_is_exporting,_but_less_than_normal'
]

manufacturing_columns_v1 = [
    'Manufacturing_On_furlough_leave',
    'Manufacturing_Working_remotely_instead_of_at_their_normal_place_of_work'
]

transport_columns_v1 = [
    'Transport_Importing_has_not_been_affected',
    'Transport_Not_been_able_to_import_in_the_last_two_weeks'
]

information_columns_v1 = [
    'Information_Not_using,_or_intending_to_use,_either_of_these_schemes',
    'Information_Job_Retention_Bonus'
]

# Step 4: Run T-Tests per sector
ttest_hospitality_v1 = perform_ttest(early_financial_period, late_financial_period, hospitality_columns_v1)
ttest_retail_v1 = perform_ttest(early_period_export, late_period_export, retail_columns_v1)
ttest_manufacturing_v1 = perform_ttest(early_period_workforce, late_period_workforce, manufacturing_columns_v1)
ttest_transport_v1 = perform_ttest(early_period_import, late_period_import, transport_columns_v1)
ttest_information_v1 = perform_ttest(early_period_strategic, late_period_strategic, information_columns_v1)

# Step 5: Display Results
print("T-Test Results: Hospitality\n", ttest_hospitality_v1, "\n")
print("T-Test Results: Retail\n", ttest_retail_v1, "\n")
print("T-Test Results: Manufacturing\n", ttest_manufacturing_v1, "\n")
print("T-Test Results: Transport and Storage\n", ttest_transport_v1, "\n")
print("T-Test Results: Information and Communication\n", ttest_information_v1, "\n")
```

```python
# Step 6: Function to interpret hypothesis test results
def interpret_results(ttest_df, sector_name, alpha=0.05):
    print(f"\nHypothesis Decisions for {sector_name}:")
    for _, row in ttest_df.iterrows():
        col = row['Column']
        p_val = row['P-Value']
        if p_val < alpha:
            print(f"  Reject null hypothesis for '{col}' (p = {p_val:.4f})")
        else:
            print(f"  Fail to reject null hypothesis for '{col}' (p = {p_val:.4f})")

# Step 7: Run hypothesis decision check per sector
interpret_results(ttest_hospitality_v1, "Hospitality")
interpret_results(ttest_retail_v1, "Retail")
interpret_results(ttest_manufacturing_v1, "Manufacturing")
interpret_results(ttest_transport_v1, "Transport and Storage")
interpret_results(ttest_information_v1, "Information and Communication")
```
